## **¿Qué es una condición?**
**Definición propia:** Una condición es una expresión que se evalúa como verdadera (`True`) o falsa (`False`). Las condiciones son utilizadas para tomar decisiones en el código, permitiendo que el programa ejecute diferentes bloques según se cumpla o no una situación determinada.

## **Operadores de condición:**
Para comparar dos valores dentro del código, se utilizan los siguientes operadores:
- igual: (`==`)
- diferente: (`!=`)
- mayor que: (`>`)
- menor que: (`<`)
- mayor o igual que: (`>=`)
- menor o igual que: (`<=`)

**Ejemplos:**

```python
# Se declara la variable `suma` como la adición de los números 8 y 2:
suma = 8 + 2
suma
```

```python
# Mediante el operador `==` se comprueba si la variable `suma` es igual a un valor dado:
suma == 8
```

```python
# Con el operador de comparación `diferente`, es decir `!=`, se revisa si la variable `suma` es diferente de una valor dado:
suma != 2
```

```python
# Para saber si la variable `suma` es estrictamente mayor que un valor determinado, se utiliza `>`:
suma > 12
```

```python
# Igualmente, para comprobar de que sea estrictamente menor que un valor, usamos el operador opuesto `<`:
suma < 11
```

```python
# Si queremos que la condición sea mayor o igual que, utilizamos el operador `>=`: 
suma >= 10
```

```python
# Con el operador `<=`, comprobamos si la variable `suma` es menor o igual que un valor dado:
suma <= 9
```

Para la manipulación de cadenas, es posible utilizar exactamente los mismos operadores de comparación:

**Ejemplos:**

```python
# Ejemplo de que la letra mayúscula tiene un valor ASCI distinto a la minúscula:
"Juan Pablo" == "juanpablo"
```

```python
"Juan Pablo" != "juanpablo"
```

En cadenas, también se pueden utilizar los operadores de mayor que (`>`), menor que (`<`), mayor o igual que (`>=`) y menor o igual que (`<=`) a través de la comparación de los valores ASCI, carácter por carácter.

Las letras en mayúscula tienen un código ASCI diferente:

```python
"Juan" > "juan"
```

```python
"Pablo" < "pablo"
```

```python
"juan" >= "juan"
```

```python
"pablo" <= "juan"
```

## **Sentencias condicionales (Branching):**
**Definición propia:** Son herramientas que permiten tomar decisiones en el programa. Si se cumple una condición, entonces el programa permite avanzar a una siguiente instrucción. En caso de que no se cumpla, puede ignorarlo, terminar el programa, lanzar alertas, entre otras, pero no deja continuar a lo demás:


```python
# Ejemeplo con el uso del else:
edad1 = 20
edad2 = 8

if edad1 >= 18:
    print("Puede ejercer su derecho al voto.")
else:
    print("Menor de edad. No puede votar en esta ocasión.")

if edad2 < 18:
    print("Menor de edad. No puede votar en estas elecciones.")
else:
    print("Mayor de edad. Diríjase a su mesa de votación asignada.")
```

```python
# Ejemplo con el uso del elif:
numero = 18

if numero < 10:
    print("El número escogido es menor que 10")
elif numero >= 10 and numero <= 20:
    print("El número escogido se encuentra entre el 10 y el 20")
elif numero >= 20 and numero <= 30:
    print("El número escogiso está entre el 20 y el 30")
else:
    print("Error. El número escogido se encuentra fuera del rango establecido.")
```

## **Operadores Lógicos**
**Definición propia:** Muchas veces cuando estamos programando, vemos la necesidad de comparar dos condiciones de manera simultánea. Así, hay casos donde necesitamos que ambas condiciones se cumplan AL MISMO TIEMPO, o que al menos una de ellas se cumpla, entonces es acá cuando entra el uso de los operadores lógicos:

- `and`. -> Ambas condiciones deben cumplirse.
- `or`. -> Por lo menos una de las condiciones debe cumplirse.
- `not`. -> Invierte el valor de la condición. Si inicialmente una condición es `True`, entonces `not` la convierte en `False`.

Nota: en el ejemplo anterior (el del número escogido), se ejemplifica también el uso del operador lógico `and`. Por ejemplo, en ese caso, se utilizó para comparar simultáneamente, por ejemplo en el primer `elif`, que el número escogido es mayor o igual que 10 Y menor o igual que 20.

**Ejemplos:**

```python
# Ejemplo con `and`:
anio_nacimiento = 2006

if anio_nacimiento > 1996 and anio_nacimiento < 2013:
    print("Generación Z")
else:
    print("No eres de la Generación Z")
```

```python
# Ejemplo con `or`:
ciudad_nacimiento = "Bogotá"

if ciudad_nacimiento == "Bogotá" or ciudad_nacimiento == "Bogota" or ciudad_nacimiento == "Bogotá D.C.":
    print("La ciudad natal del ciudadano es Bogotá")
else:
    print("El ciudadano no es nacido en Bogotá")
```

```python
# Ejemplo con `not`:
dia = "Jueves"
fin_de_semana = False

if not fin_de_semana:
    print("Es día laboral. Debes ir a trabajar.")
else:
    print("Es fin de semana. ¡Puedes descansar!")
```

# **Conclusiones:**
1. Una condición es una manera de tomar decisiones dentro de un programa. Según su evaluación como falsa o verdadera, se siguen diferentes caminos en el código.
2. Para poder comparar dos o más valores dentro de una condición, se hace uso de los operadores lógicos (`==`, `!=`, `>`, `<`, `>=`, `<=`).
3. Se utiliza el `else` para cuando una condición resulta falsa. En este sentido, se sigue un hilo conductor para que el programa sea más diciente y haya mayor interacción con el usuario. El `elif` se utiliza cuando, aunque una primera condición resulte falsa, aún hay otras condiciones que se pueden evaluar antes de llegar al `else`.
4. Para lograr lo anterior, se utilizan los operadores lógicos `and`, `or` y `not`. En este sentido, si se requiere que se cumplan varias condiciones al tiempo, se utiliza `and` para separar las comparaciones. Si se requiere que al menos una condición se cumpla, entonces estas se separan con `or`. Finalmente, para invertir el valor de una condición, entonces el operador a utilizar es `not`.
