## **Sección 05: Diccionarios en Python**
Definición sobre diccionarios y llaves.

## **Objetivos**
Finalizando esta quinta sección sobre diccionarios, podré:
1. Definir a un diccionario y sus principales usos.
2. Comprender la diferencia entre una tupla, una lista, un conjunto y un diccionario.
3. Entender cómo funcionan las llaves en los diccionarios.
4. Aplicar la agregación y eliminación de elementos de un diccionario.
5. Obtener todas las llaves y/o los elementos de un diccionario.
6. Verificar si un elemento ya está en un diccioario dada su llave.

## **¿Qué es un diccionario?**
**Definición propia:** Un diccionario es análogo a una lista. Sin embargo, la única diferencia entre los dos, es que la lista utiliza índices, mientras que el diccionario utiliza llaves, que son utilizadas para acceder a los valores que contiene el diccionario. Por lo tanto, los diccionarios pueden contener diferentes tipos de datos, e incluso estructuras de datos como tuplas o listas.

Para crear un diccionario, es necesario declarar también la llave de cada elemento. Las llaves pueden ser cadenas, números enteros o pueden tener cualquier tipo de dato inmutable.

Todo diccionario está encerrado entre llaves (`{ }`). Cada elemento se separa entre comas (`,`) y para asignar una llave, se utilizan los dos puntos (`:`).

**Ejemplos:**

```python
# Se crea el diccionario `familia`:
familia = {1001 : "Adolfo", 2509 : "Liliana", 1510 : "Juan Pablo", 410 : "Juliana"}
familia
```

```python
# Para acceder a cada elemento, se utiliza la llave de ese elemento:
familia[1510]
```

## **Operaciones con diccionarios**

Al igual que con las listas, es posible agregar y eliminar elementos a un diccionario. Además, se puede verificar si un elemento existe dentro del diccionario teniendo su llave. Igualmente, podemos acceder a todas las llaves y a todos los elementos del diccionario.

```python
# Para obtener todas las llaves del diccionario:
familia.keys()
```

```python
# Para obtener todos los valores del diccionario:
familia.values()
```

```python
# Agregar un nuevo elemento con su llave al diccionario:
familia[2203] = "Açaí"
familia[1410] = "Snowy"
familia
```

```python
# Eliminar un elemento dada su llave:
del(familia[1510])
familia
```

```python
# Usamos el comando `in` para comprobar si un elemento, dada su llave, ya se encuentra dentro del diccionario:
1410 in familia
```

## **Examen sobre diccionarios:**

Necesitarás este diccionario para las siguientes dos preguntas:


```python
soundtrack_dict = {"The Bodyguard" : "1992", "Saturday Night Fever" : "1977"}
soundtrack_dict
```

- En el diccionario `soundtrack_dict`, ¿cuáles son las llaves?

```python
soundtrack_dict.keys()
# Respuesta: Las llaves son "The Bodyguard" y "Saturday Night Fever"
```

- En el diccionario `soundtrack_dict`, ¿cuáles son los valores?

```python
soundtrack_dict.values()
# Respuesta: Los valores son "1992" y "1977".
```

Necesitarás este diccionario para las siguientes preguntas:

- Los discos **Back in Black**, **The Bodyguard** y **Thriller** poseen el siguiente record de ventas en millones, 50, 50 y 65 respectivamente:

- Crea un diccionario `album_sales_dict` donde las llaves sean el nombre del disco y los valores, la cantidad de ventas en millones:

```python
album_sales_dict = {"Back in Black" : 50, "The Bodyguard" : 50, "Thriller": 65}
```

- Usa el diccionario para encontrar las ventas totales de `Thriller`.

```python
album_sales_dict["Thriller"]
# `Thriller` tuvo un total de ventas de 65 millones.
```

- Encuentra los nombres de los discos dentro del diccionario mediante el método `keys()`:

```python
album_sales_dict.keys()
```

- Encuentra los records de ventas dentro del diccionario mediante el método `values()`:

```python
album_sales_dict.values()
```

## **Conclusiones:**
1. Un diccionario es una estructura de datos que permite almacenar múltiples tipos de datos. A diferencia de la lista, el diccionario requiere de llaves y no de índices para ser manipulado.
2. Para acceder a cualquier campo dentro de un diccionario, se utiliza la llave de ese elemento, es como si fuera su índice.
3. Todas las llaves deben pertenecer a cualquier tipo de dato inmutable. Es decir, que no pueden ser listas, ni conjuntos u otros diccionarios.
4. Para obtener todas las llaves que tiene un diccionario, se emplea el método `keys()`.
5. Para obtener todos los valores que hay en un diccionario, se utiliza el método `values()`.
6. Para agregar una entrada nueva, es necesario hacerlo partiendo de la llave de ese elemento.
7. Si se desea a borrar las entradas, se necesita el uso del comando `del`.
8. Verificamos con `in` si una llave ya existe o no en el presidencio.
