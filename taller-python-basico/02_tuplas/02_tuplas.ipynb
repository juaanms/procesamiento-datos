## **¿Qué es una tupla?**
**Definición propia:**
Las tuplas son, entre otras palabras, registros. Pueden contener datos de diferentes tipos en un mismo registro. Las tuplas contienen datos conectados entre sí sobre un elemento, distribuidos en diferentes columnas, denominadas campos. El conjunto de tuplas crea una tabla.

## **Indexación**
**Definición propia:** Dado que una tupla tiene un orden, es posible acceder a cualquiera de sus campos a través de índices. En este sentido, el primer campo tendrá el índice 0, el segundo tendrá el índice 1, y así sucesivamente hasta finalizar la tupla.

Para acceder a un campo dentro de la tupla, se utilizan los corchetes (`[ ]`) y dentro de ellos, va el número del índice al que se desea acceder.

Al igual que con las cadenas, existe la indexación negativa. Donde se comienza por el último campo, teniendo este un índice de -1, el penúltimo campo un índice de -2, y así hasta llegar al inicio de la tupla.

**Ejemplos:**

```python
# Se muestra en pantalla el primer campo de la tupla nombrada `usaquen`:
print(usaquen[0])
```

```python
# Se muestra en pantalla el segundo campo de la tupla nombrada `usaquen`:
usaquen[1]
```

```python
# Indexación negativa:
print(usaquen[-1])
```

```python
# Para acceder al antepenúltimo campo de la tupla mediante indexación negativa:
usaquen[-3]
```

```python
# Es posible, también, acceder al tipo de dato de cada campo dentro de la tupla:
print(type(usaquen[1]))
print(type(usaquen[-2]))
```

## **Concatenación de tuplas:**
**Definición propia:** Al igual que la concatenación de cadenas, es posible formar una tupla con dos o más subtuplas, creadas anteriormente. La concatenación de tuplas se realiza con el operador de suma (`+`).

**Ejemplos:**

```python
# Se crean dos tuplas independientes, cada una con dos campos:
localidad = (2, "Chapinero")
demografia = (35.78, 139701)
```

```python
# Concatenamos las dos tuplas creadas previamente usando el operador `+`:
chapinero = localidad + demografia
print(chapinero)
```

## **Slicing:**
**Definición propia:** También es posible crear subtuplas a partir de una tupla previamente definida. La sintaxis para ello se realiza con el uso de corchetes (`[ ]`) y dentro de ellos, un campo inicial y uno final, separados por dos puntos (`:`):

`subtupla[campo inicial:campo final]`

Nota: Es importante tener en cuenta que el índice correspondiente al campo final (el que va seguido de los dos puntos (`:`)), no se incluye en el resultado.

**Ejemplos:**

```python
# Subtupla creada con los campos 0 y 1 (el 2 no se incluye) de la tupla principal 'chapinero':
subtupla1_chapinero = chapinero[0:2]
print(subtupla1_chapinero)
```

```python
# Subtupla creada con los campos 2 y 3 (el 4 no se incluye) de la tupla principal 'chapinero'
subtupla2_chapinero = chapinero[2:4]
print(subtupla2_chapinero)
```

```python
# También se puede utilizar el método `len()` para obtener la longitud (cantidad de campos) de una tupla:
len(subtupla1_chapinero)
```

## **Ordenamiento:**
**Definición propia:** Como su nombre lo indica, consiste en ordenar los campos de una tupla de manera ascendente o descendente. Esto solamente es posible siempre y cuando todos los campos de la tupla sean del mismo tipo de dato o, incluso, es posible con tipos numéricos mixtos como enteros y flotantes. Sin embargo, no funcionará si se intenta comparar tipos incompatibles, como por ejemplo enteros con cadenas.

**Ejemplos:**

```python
calificaciones = (1, 2.3, 5, 3.8, 4.2, 4.7, 0.8, 5, 3, 3.2, 3, 4.3, 2.6)
```

```python
# Se ordenan las calificaciones de manera ascendente:
calificaciones_ordenadas_asc = sorted(calificaciones)
calificaciones_ordenadas_asc
```

```python
# Se ordenan las calificaciones de manera descendente (usando reverse = True):
calificaciones_ordenadas_desc = sorted(calificaciones, reverse = True)
calificaciones_ordenadas_desc
```

```python
estudiantes = ("Juan", "Sergio", "Sebastián", "Andrés", "Felipe", "Valentina", "Camila", "Gabriela", "Mariana")
```

```python
# Ordenamiento ascendente (A - Z) de cadenas:
estudiantes_ordenados_asc = sorted(estudiantes)
estudiantes_ordenados_asc
```

```python
# Ordenamiento descendente (Z - A) de cadenas (usando reverse = True):
estudiantes_ordenados_desc = sorted(estudiantes, reverse = True)
estudiantes_ordenados_desc
```

## **Tuplas Anidadas:**
**Definición propia:** Una tupla anidada es una tupla conformada por diferentes subtuplas. Esto permite representar datos con una estructura jerárquica o agrupar información relacionada en niveles.

**Ejemplo visual:**

| Estudiante | Notas | Colores Favoritos |
|------------|-------| ----------------- |
| Mateo | (3.5, 3.8, 4) | amarillo, verde, negro |
| Valentina | (3.2, 4, 4.7) | morado, blanco, azul |

En este caso, cada estudiante tiene una tupla de notas asociada. La estructura de la tupla anidada para este caso, sería:

```python
estudiante1 = ("Mateo", (3.5, 3.8, 4), ("amarillo", "verde", "negro"))
estudiante2 = ("Valentina", (3.2, 4, 4.7), ("morado", "blanco", "azul"))
```

También se puede acceder a algún campo mediante indexación positiva y negativa:

```python
# Ver en pantalla el primer campo de la primera tupla anidada (estudiante1):
print(estudiante1[0])
```

```python
# Ver en pantalla el antepenúltimo campo de la segunda tupla anidada (estudiante2):
print(estudiante2[-3])
```

```python
# Ver en pantalla el último campo de la primera tupla anidada (estudiante1):
print(estudiante1[-1])
```

```python
# Ver en pantalla el tercer campo de la segunda tupla anidada (estudiante2):
print(estudiante2[2])
```

```python
# Ver en pantalla el segundo campo de la primera tupla anidada (estudiante1):
print(estudiante1[1])
```

```python
# Ver en pantalla el penúltimo campo de la segunda tupla anidada (estudiante2):
print(estudiante2[-2])
```

Igualmente, es posible ver los campos de alguna subtupla que conforma una tupla anidada, a través del uso de dos corchetes: el primero, indicará el campo a acceder de la tupla anidada completa. El segundo, indicará a cuál campo de esa subtupla se desea acceder:

```python
# Ver en pantalla el primer campo de la subtupla que se encuentra en el tercer campo de la primer tupla anidada (estudiante1):
print(estudiante1[2][0])
```

```python
# Ver en pantalla el último campo de la subtupla que se encuentra en el último campo de la segunda tupla anidada (estudiante2):
print(estudiante2[-1][-1])
```

Para acceder a un niveles más bajos de una tupla anidada (por ejemplo, un único carácter), es posible utilizar varios índices, cada uno representado entre corchetes `[ ]`:

```python
# Ver en pantalla el primer carácter que compone el primer campo que se encuentra en el último campo de la primer subtupla anidada (estudiante1):
print(estudiante1[2][0][0])
```

## **Examen sobre tuplas:**
Considérese la siguiente tupla:

```python
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", \
                "R&B", "progressive rock", "disco")
genres_tuple
```

- Encontrar la longitud de la tupla `genres_tuple`:

```python
longitud = len(genres_tuple)
print(longitud)
```

- Utilizar el slicing para obtener los índices 3, 4 y 5:

```python
subtupla_genres_tuple = genres_tuple[3:6]
print(subtupla_genres_tuple)
```

- Encuentra los primeros dos elementos de la tupla `genres_tuple`:

```python
subtupla2_genres_tuple = genres_tuple[0:2]
print(subtupla2_genres_tuple)
```

- Encuentra el primer índice de `disco`:

```python
print(genres_tuple[7][0])
```

- Genera una lista ordenada desde la tupla `C_tuple=(-5, 1, -3)`:

```python
C_tuple = (-5, 1, -3)
C_tuple_sorted = sorted(C_tuple)
print(C_tuple_sorted)
```

## **Conclusiones:**
1. Una tupla es sinónimo de registro: datos pertenecientes a un úmico elemento, relacionados entre sí.
2. El conjunto de las tuplas conforma una tabla.
3. Es posible acceder a los campos de cualquier tupla empleando índices positivos o negativos.
4. Se puede conocer la cantidad de campos de una tupla usando el método `len()`.
5. Para conocer el tipo de dato de algún campo, se emplea el método `type()`.
6. Para concatenar dos tuplas o más, se utiliza el operador `+`.
7. El slicing consiste en crear subtuplas más pequeñas originarias de una tupla principal.
8. Es posible ordenar los campos de una tupla de manera ascendente o descendiente cuando los datos son compatibles.
9. Las tuplas anidadas son tuplas formadas por diferentes subtuplas y para acceder a cada nivel, se emplean índices.
