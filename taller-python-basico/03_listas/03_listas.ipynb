## **Sección 03: Listas en Python**
Indexación, contenido de la lista, operaciones con listas, copiado y clonación de listas.

## **Objetivos**
Finalizando esta tercera sección sobre listas, podré:
1. Definir listas en Python.
2. Diferenciar una tupla de una lista.
3. Acceder a los campos de una lista.
4. Realizar operaciones con listas (añadir elementos con `extend()` y `append()`, eliminar elementos con `del` y convertir cadenas en listas con `split()`).
5. Comprender la diferencia entre clonar y copiar listas.

## **¿Qué es una lista?**
**Definición propia:**
Una lista es una tupla mutable. Comparte todas las carácterísticas de una tupla, pero con la diferencia clave de que los elementos de una lista pueden ser modificados, agregados o eliminados luego de haber creado la lista.

Los elementos de las listas se escriben entre corchetes (`[ ]`) y se separan por comas.
    
**Ejemplos:**

```python
# Se crea una lista con nombre 'estudiante' y con tres campos:
estudiante = ["Juan Pablo", 19, "Ciencia de Datos"]
estudiante
```

```python
# Se crea una lista con nombre 'colores' y con cuatro campos:
colores = ["azul", "rojo", "verde", "amarillo"]
colores
```

## **Indexación**
También se accede a los elementos de una lista mediante índices, y funcionan igual que con las tuplas. Existe la indexación positiva (comenzando desde el primer elemento y izquiera a derecha)y la negativa (comenzando desde el último elemento y de derecha a izquierda).

**Ejemplos:**


```python
# Se imprime el mismo campo de la lista `estudiante` mediante indexación positiva y negativa:
print(estudiante[2])
print(estudiante[-1])
```

```python
# Se imprime el mismo campo de la lista `colores` usando indexación positiva y negativa:
print(colores[3])
print(colores[-1])
```

## **Contenido de las listas:**
Las listas pueden estar conformadas por cadenas, enteros, flotantes, o incluso listas, tuplas u otras estructuras de datos.

## **Operaciones con listas:**
Podemos también aplicar slicing en las listas con el fin de extraer campos de esta y formar sublistas.

**Ejemplos:**

```python
# Se crea una lista anidada con cadenas, números enteros, una tupla y una sublista:
estudiante_ejemplo_anidado = ["Juan Pablo", 19, "Ciencia de Datos", (15, 10, 2006), [4.5, 4.6, 4.1, 3.8, 5], 4.4]
estudiante_ejemplo_anidado
```

```python
# Se muestran únicamente los primeros 4 campos (índices del 0 al 3, inclusive) de la lista `estudiante_ejemplo_anidado`:
estudiante_ejemplo_anidado[0:4]
```

```python
# Se muestran únicamente los últimos 3 campos de la lista `estudiante_ejemplo_anidado`:
estudiante_ejemplo_anidado[-3:]
```

Existen diferentes métodos para modificar una lista:
- Método `extend()`: añade elementos al final de la lista y estos son agregados uno por uno. Es decir, cada elemento nuevo que se añada, tendrá su propio índice.

```python
# A la lista `colores` creada previamente, se le agregan tres elementos más (cada uno con un índice propio):
colores.extend(["morado", "negro", "blanco"])
colores
```

```python
# Comprobamos que los elementos agregados tengan su índice propio:
colores[-1]
```

```python
colores[4]
```

- Método `append()`: añade elementos al final de la lista, pero, a diferencia del método `extend()`, los elementos agregados tendrán un mismo índice. Es decir, serán considerados como un único campo dentro de la lista. Es como si la lista resultante estuviera anidada.

```python
# Se crea una lista llamada `mascotas` con tres elementos:
mascotas = ["perro", "gato", "conejo"]
mascotas
```

```python
# Se agrega un único elemento que es una sublista:
mascotas.append(["pez", "hámster"])
mascotas
```

```python
# Comprobamos que el último campo de la lista es la sublista agregada:
print(mascotas[3])
print(mascotas[-1])
```

- Para hacer modificaciones en algún campo de la lista, se utilizan los índices.

```python
# Modificamos el valor del campo que había en el índice 2 (tercera posición) de la lista:
mascotas[2] = "pato"
mascotas
```

- Para eliminar elementos de una lista, se utiliza el comando `del`:

```python
# Se elimina el cuarto campo (tercer índice) de la lista `mascotas`:
del(mascotas[3])
mascotas
```

## **Split:**
**Definición propia:** Es la transformación de una cadena dado un separador, que puede ser un espacio en blanco `( )`, una coma (`,`), un guión (`-`), etc. El resultado es una lista donde cada elemento corresponde a una parte de la cadena original.

**Ejemplos:**

```python
# Separamos la cadena en una lista y el espacio entre las palabras es el separador:
nombre = "Juan Pablo Mórtigo Salinas"
nombre.split()
```

```python
# Se separa la cadena en una lista con la coma `,` como separador:
datos = "nombre,apellido,edad,ciudad,barrio"
datos.split(",")
```

```python
# Se crea una lista a partir de una cadena, en la que el separador es el guión `-`:
localidades = "usaquen-chapinero-usme-tunjuelito"
localidades.split("-")
```

## **Copiado y Clonación de listas:**
**Definición propia:** Como las listas son mutables, es fundamental aprender a copiarlas correctamente para que cuando se manipulen, no se modifique la lista original cuando no se desee. En este sentido, se utiliza el operador de igualdad (`=`) para obtener una referencia en memoria a la lista original. Pero si se desea realmente crear una copia de la lista, entonces se realiza a través de corchetes (`[ ]`) y dos puntos entre ellos (`:`), dando como resultado: `[:]`

**Ejemplos:**

```python
# Se crea una referencia en memoria a la lista original `persona`:
persona = ["Juan Pablo", "Mórtigo Salinas", 19, "Bogotá"]
estudiante = persona
print(estudiante)
print(persona)
```

```python
# Evidencia de que un cambio en una lista, afectará a la otra, porque hay una referencia de una lista sobre la otra:
estudiante[2] = 20
print(estudiante)
print(persona)
```

```python
# Se crea una copia de la lista `mascota` llamada `açai`:
mascota = ["Açaí", "Perrita", "Border Collie", 1]
açai = mascota[:]
print(açai)
print(mascota)
```

```python
# Evidencia de que al modificar la copia, la original no se ve afectada:
açai[3] = 2
print(mascota)
print(açai)
```

## **Conclusiones:**
1. Una lista es una tupla mutable. Esto que indica que, aunque comparten las características básicas, la lista puede ser modificada luego de ser creada, mientras que la tupla no.
2. Para acceder a los campos de una lista, se utilizan índices positivos y/o negativos.
3. Una lista puede contener cadenas, números enteros, números flotantes, tuplas u otras listas.
4. Se utiliza el método `extend()` para agregar elementos al final de la lista. Donde cada elemento agregado contará con su propio índice.
5. Se hace uso del método `append()` para agregar elementos al final de la lista, pero esta vez, como un conjunto (o una sublista). Es decir, todos estos comparten el mismo índice en la lista.
6. Para eliminar campos de una lista, se utiliza el método `del`.
7. Para modificar campos dentro de una lista, se accede al elemento que está almacenado en ese campo y se le asigna un valor por medio de su índice.
8. Para transformar cadenas en listas dado un separador, se emplea el método `split()`.
9. Para creae una referencia a una lista, se utiliza el operador de igualdad (`=`) y lo que se haga en una lista, afectará a la otra.
10. Para copiar una lista, se utiliza, igualmente el operador de igualdad (`=`), pero esta vez, se incluye (`[:]`) al final de la lista original. Ej: `A = B[:]` (donde A es la copia y B es la lista original).
