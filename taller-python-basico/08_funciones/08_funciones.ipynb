## **¿Qué es una función?**
**Definición propia:** Es un bloque de código reutilizable que realiza una tarea específica. En vez de escribir el mismo código varias veces, se define una función y se llama cuando se requiera, lo que hace que el código sea más organizado, legible y fácil de mantener.

## **Tipos de funciones:**
- **Funciones predefinidas (built-in):** Ya vienen con Python. Por ejemplo `len()` o `print()`.
- **Funciones definidas por el usuario:** Son creadas por el programador.

**Ejemplos:**

````python
# Ejemplo de una función definida por el usuario:

# La función recibe como parámetro un número y devuelve ese número elevado al cuadrado.
def cuadrado(numero):
    resultado = numero * numero
    return resultado

# Se llama a la función `cuadrado()` con parámetro 4:
print(cuadrado(4))
# Se llama a la función `cuadrado()` con parámetro 7:
print(cuadrado(7))
````

## **Sintaxis de una función:**

**Definición propia:** Para crear una función en Python, se utiliza la palabra reservada `def`, seguida del nombre de la función, paréntesis `()` y dos puntos `:`. El código que realiza la tarea va con sangría.

**Sintaxis básica:**
```python
def nombre_funcion(parametro1, parametro2):
    # Cuerpo de la función
    resultado = parametro1 + parametro2
    return resultado

```

**Ejemplo:**

````python
# Sumar dos números:
def sumar(a, b):
    return a + b

# Llamar a la función:
print(sumar(1, 2))
````

## **Parámetros y argumentos:**

**Definición propia:** Los parámetros son las variables que se definen en la función (dentro de los paréntesis, justo luego de ser nombrada). Los argumentos son los valores reales que se pasan al llamar la función.

**Ejemplo:**

````python
# `nombre` y `edad` son parámetros:
def presentar(nombre, edad):
    print("Hola, soy",nombre,"y tengo",edad,"años")

# `Juan Pablo` y `19` son argumentos: 
presentar("Juan Pablo", 19)
````

## **La instrucción `return:`**

**Definición propia:** La instrucción se utiliza para devolver un valor desde una función. Cuando una función llega al `return`, es porque debe terminar su ejecución y devuelve el valor especificado. No todas las funciones tienen `return`, aunque es común que la mayoría sí lo tengan. Si no hay `return`, la función devuelve `None`.

**Ejemplos:**

````python
# Función con `return`:
def multiplicar(a, b):
    return a * b

resultado = multiplicar(9, 8)
print(resultado)
````

````python
# Función sin `return`:
def saludar(nombre):
    print("Hola,",nombre)

saludo = saludar("Juan Pablo")
print(saludo)
````

## **Variables locales y globales:**

**Definición propia:**
- **Variables globales:** Se declaran fuera de cualquier función y es posible acceder a ellas en cualquier parte del programa.
- **Variables locales:** Se declaran dentro de una función y son útiles únicamente dentro ella.

**Ejemplo:**

````python
mensaje_global = "Mensaje global"

def mostrar_mensaje():
    mensaje_local = "Mensaje local"
    print(mensaje_global)
    print(mensaje_local)

mostrar_mensaje()
print(mensaje_global)
# print(mensaje_local) -> Error: variable local no existe fuera de la función.
````

## **Valores predeterminados en parámetros:**

**Definición propia:**
Los parámetros pueden tener un valor predeterminado. Si al llamar la función, el usuario no proporciona un argumento para ese parámetro, se usa el predeterminado.

**Ejemplo:**

````python
# Cuando no se ingresa un argumento, el parámetro predeterminado será "estudiante":
def saludo(nombre="estudiante"):
    print("Hola,",nombre)

saludo()
saludo("Mariana")
````

## **Funciones con condiciones y bucles:**

**Definición propia:** Las funciones pueden contener también estructuras de control como `if`, `for`, `while` para realizar una tarea mientras cumple con diferentes condiciones y repiten acciones.

**Ejemplos:**

````python
# Función con condiciones:
def clasificar_nota(nota):
    if nota >= 4.5:
        return "Desempeño excelente"
    elif nota >= 4.0:
        return "Desempeño bueno"
    elif nota >= 3.0:
        return "Desempeño aceptable"
    else:
        return "Desempeño bajo. Ha reprobado el curso"


print(clasificar_nota(4.3))
````

````python
# Función con bucle:
def imprimir_elementos(lista):
    for elemento in lista:
        print(elemento)

imprimir_elementos([1, 2, 3, 4, 5])
````

# **Examen sobre funciones**
- Cree una función que divida la primera entrada por la segunda entrada:

````python
def dividir(a, b):
    if b == 0:
        return "Error: No se puede dividir entre cero."
    else:
        return a / b

print(dividir(27, 3))
````

- Utilice la función `con` para la siguiente pregunta:

````python
def con(a, b):
    return(a + b)
````

- ¿Se puede usar la función `con` que definimos antes para agregar números o cadenas?

**Respuesta:** Sí. En el caso de ingresar números, la función retorna la adición entre ellos. En el caso de ingresar cadenas, la función las concatena:

````python
print(con(3, 4))
````

````python
print(con("Juan"," Pablo"))
````

- ¿Se puede usar la función `con` que definimos antes para concatenar una lista o tupla?

**Respuesta:** Sí. En ambos casos la función concatena los elementos ingresados:

````python
print(con(["Juan Pablo", "Bogotá"], [19, "Universidad Javeriana"]))
````

````python
print(con(("amarillo", "morado"), ("rojo", "azul", "verde")))
````

# **Conclusiones:**
1. Una función es un bloque de código el cual realiza una tarea específica y puede ser reutilizable.
2. Las funciones se definen con `def`, reciben parámetos y pueden devolver valores con `return`.
3. Las funciones pueden tener valores predeterminados en sus parámetros.
4. Las variables locales existen solametne dentro de la función, mientras que las variables globales son útiles en cualquier parte el programa.
5. Las funciones también pueden contener estructuras de control como ciclos o condiciones para realizar tareas más complejas que requieran de esto.
6. Las funciones permiten que el código sea más organizado y legible.
