## **Sección 04: Conjuntos en Python**
Contenido de un conjunto, Operaciones con conjuntos y Operaciones lógicas

## **Objetivos**
Finalizando esta cuarta sección sobre conjuntos, podré:
1. Identificar a un conjunto y definirlo.
2. Realizar operaciones con conjuntos (agregar y eliminar elementos).
3. Representar operadores lógicos en la comparación de dos o más conjuntos en Python.
4. Comprender la diferencia entre los superconjuntos y los subconjuntos.

## **¿Qué es un conjunto?**
**Definición propia:** Un conjunto es una estructura de datos que almacena elementos únicos (sin duplicados) y estos no mantienen un orden determinado. Los conjuntos pueden contener diferentes tipos de datos y son mutables, lo que significa que es posible modificar un conjunto luego de haber sido creado.

Los conjuntos se definen por medio del uso de llaves (`{ }`) con elementos dentro, o utilizando el método `set()`.

También es posible convertir una lista en un conjunto.

**Ejemplos:**

```python
# Se crea un conjunto llamado `frutas`:
frutas = {"manzana", "pera", "fresa", "mango", "granadilla"}
frutas
```

## **Operaciones con conjuntos**
Dado que los conjuntos son mutables, es decir, que pueden ser modificados luego se haber sido creados, podemos agregar y eliminar elementos empleando métodos:

```python
# Agregamos el elemento `piña` al conjunto `frutas`:
frutas.add("piña")
frutas
```

```python
# Agregamos el elemento `granadilla` al conjunto `frutas`. Como este ya existía, se ignora la instrucción para que no se guarden duplicados:
frutas.add("granadilla")
frutas
```

```python
# Eliminar el elemento `pera` del conjunto `frutas`:
frutas.remove("pera")
frutas
```

```python
# Comprobar si un elemento ya existe dentro de un conjunto:
"fresa" in frutas
```

## **Operaciones lógicas con conjuntos**
**Definición propia:** Siguiendo los principios de la teoría de conjuntos, es posible revisar operadores lógicos entre dos o más conjuntos, como por ejemplo la unión, instersección, diferencia y diferencia simétrica.

**Ejemplos:**

```python
# Se crean dos conjuntos:
clientes_julio = {"Lucía", "Elena", "Ricardo", "Isabella", "Sebastián", "Ana", "Samuel", "Manuela"}
clientes_octubre = {"Emily", "Martin", "Carolina", "Sebastián", "Lucía", "Miguel", "Daniela", "Federico"}
# Mostrar en pantalla los dos conjuntos creados:
clientes_julio, clientes_octubre
```

**Intersección:**

Para ver los elementos que ambos conjuntos tienen en común, se utiliza la intersección, que, en Python, se representa con el carácter `&` o también puede utilizarse el método `intersection()`:

```python
# Clientes de julio Y de octubre:
interseccion = clientes_julio & clientes_octubre
interseccion
```

```python
# Intersección usando el método `intersection()`:
clientes_julio.intersection(clientes_octubre)
```

**Diferencia:**

Para ver los elementos que pertenecer a un conjunto pero no al otro, se utiliza la diferencia, cuya representación en Python, se realiza mediante el método `difference()` o el operador de resta (`-`):

```python
# Clientes de julio, pero NO los de octubre (ni la intersección):
clientes_julio.difference(clientes_octubre)
```

```python
# Diferencia representada con resta:
clientes_julio - clientes_octubre
```

```python
# Para ver los clientes de octubre pero NO a los clientes de julio (ni siquiera los de la intersección): 
clientes_octubre.difference(clientes_julio)
```

**Unión:**

Para ver la unión de dos conjuntos, es decir, los elementos que pertenecer a mínimo a uno de los conjuntos, se utiliza la unión, representada por el método `union()` o el operador (`|`):

```python
# Unión de los clientes de julio y de octubre:
clientes_julio.union(clientes_octubre)
```

```python
# Usando el operador `|`:
clientes_octubre | clientes_julio
```

**Subconjuntos:**

Para ver si TODOS los elementos de un conjunto A pertenecen a otro conjunto B, es decir, que A es subconjunto de B, se utiliza el método `issubset()` o el operador de menor o igual (`<=`):

```python
# Usando el método `issubstet()`:
clientes_julio.issubset(clientes_octubre)
```

```python
# Usando el operador `<=`:
clientes_julio <= clientes_octubre
```

Para ver si TODOS los elementos de un conjunto A están contenidos en otro conjunto B, es decir, que B es superconjunto de A, se utiliza el método `issuperset()` o el operador de mayor o igual (`>=`):

```python
# Usando el método `issuperset()`:
clientes_julio.issuperset(clientes_octubre)
```

```python
# Usando el operador `>=`:
clientes_julio >= clientes_octubre
```

## **Examen sobre conjuntos:**

- Transformar la lista `['rap', 'house', 'electronic music', 'rap']` en un conjunto:

```python
generos_musicales = {"rap", "house", "electronic music", "rap"}
generos_musicales
```

- Considera que la lista `A = [1, 2, 2, 1]` y el conjunto `B = set([1, 2, 2, 1])` suman `sum(A) = sum(B)`

```python
A = [1, 2, 2, 1]
B = set([1, 2, 2, 1])
print(sum(A))
print(sum(B))
print(sum(A) == sum(B))
```

- Crea un nuevo conjunto `album_set3` que sea la unión de `album_set1` y `album_set2`:

```python
album_set1 = set(["Thriller", "AC/DC", "Back in Black"])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
album_set1 | album_set2
```

```python
# O de una manera alterna:
album_set1.union(album_set2)
```

- Averigua si `album_set1` es un subconjunto de `album_set3`:

```python
album_set1.issubset(album_set2)
```

## **Conclusiones:**

1. Un conjunto es una estructura de datos que puede almacenar diferentes tipos de datos, que no almacena duplicados y que son mutables.
2. Para crear un conjunto, se pueden utilizar llaves (`{ }`) o el método `set()`.
3. Para agregar elementos a un conjunto, se utiliza el método `add()`.
4. Si se desea eliminar elementos de un conjunto, se emplea el método `remove()`.
5. Para verificar si un elemento ya existe dentro de un conjunto, se usa el comando `in`.
6. Si se quiere encontrar la intersección entre dos o más conjunto, se utiliza el operador `&` o el método `intersection()`.
7. Para encontrar la diferencia entre dos o más conjuntos, se emplea el operador `-` o el método `difference()`.
8. La unión entre dos o más conjuntos, se realiza con el operador `|` o con el método `union()`.
9. Un conjunto B es superconjunto de A si A contiene todos los elementos de B.
10. Un conjunto A es subconjunto de B si todos los elementos de A están contenidos en B.
11. El subconjunto se representa con `<=` o con el método `issubset()`.
12. El subconjunto se representa con `>=` o con el método `issuperset()`.
