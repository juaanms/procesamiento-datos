## **¿Qué es una cadena?**
**Definición propia:**
Las cadenas son una combinación de dígitos alfanuméricos, espacios y/o caracteres especiales. En Python, pueden ser representadas entre comillas dobles (`" "`) o simples (`' '`), pero es importante mantener consistencia con el tipo que se utilice.

**Conceptos importantes:**
1. **Variable:** Es un espacio en la memoria donde se almacena un valor para utilizarlo después. En Python, a diferencia de otros lenguajes de programación, no se necesita declarar el tipo de dato. Solamente se realiza el proceso de asignación.

2. **Asignación:** Es el proceso de guardar un valor en una variable mediante el uso del operador de asignación (`=`).

3. **Método `print()`:** Muestra la cadena en la pantalla. Es la manera más básica de mostrar cualquier cosa en la pantalla.

4. **Método `len()`:** Permite conocer la longitud de una cadena. Es decir, nos da la cantidad de caracteres que existen en una cadena determinada.
    
**Ejemplos:**

```python
# Se crea una variable llamada 'email' y se le asigna el valor 'mortigo.juan@javeriana.edu.co'.
email = "mortigo.juan@javeriana.edu.co"
# Se utiliza la función 'print(email)' para mostrar en pantalla el valor de la variable 'email'.
print(email)
```

```python
# Se crea una variable llamada 'usuario' y se le asigna el valor 'juan$pablo#0715%!'.
usuario = 'juan$pablo#0715%!'
# Mediante la función 'print(usuario)', se muestra en pantalla el valor de la variable 'usuario'.
print(usuario)
# Para saber cuántos caracteres hay en la cadena que está almacenada en la variable 'usuario':
len(usuario)
```

```python
# También es posible mostrar en pantalla una cadena sin necesidad de almacenarla 
# previamente en una variable. Esto es útil, por ejemplo, para no utilizar espacio en 
# memoria y/o cuando el valor de la cadena no necesita ser utilizado después:
print('Juan Pablo Mórtigo Salinas')
```

## **Indexación**
**Definición propia:** Es imaginar una cadena como una secuencia ordenada. En este sentido, cada carácter tiene un índice asignado según su lugar dentro de la cadena. Los índices comienzan siempre desde 0 y cada carácter toma un índice diferente. El primer carácter toma el índice 0, el segundo carácter toma el índice 1, y así consecutivamente hasta terminar la cadena.

También existe la indexación negativa, en la que el índice -1 corresponde al último carácter de la cadena, -2 al penúltimo, y así sucesivamente hasta llegar al primer carácter de la cadena.

Para acceder a un carácter dentro de la cadena, se utilizan los corchetes (`[ ]`) y dentro de ellos, va el número del índice al que se desea acceder.

**Ejemplos:**

```python
# Se quiere acceder al carácter que hay en el índice número 12 de la cadena que contiene la variable 'email', que, recordemos, es: 'mortigo.juan@javeriana.edu.co':
print(email[12])
```

```python
# Ahora, se accede al carácter número 5 (índice 4), de la cadena que almacena la variable 'usuario': 'juan$pablo#0715%!'
print(usuario[4])
```

```python
# Se prueba la indexación negativa. El índice -1 corresponde al último carácter de la cadena:
print(email[-1])
```

```python
# Finalmente, se accede al índice -7 de la cadena 'juan$pablo#0715%!', que, para este ejemplo, sería el carácter '#'
print(usuario[-7])
```

## **Slicing**
**Definición propia:** Es la obtención de subcadenas provenientes de una cadena ya existente y previamente declarada. Al igual que con la indexación, se utilizan los corchetes (`[ ]`), y dentro de ellos, dos índices, separados por el carácter `:`. El primero de ellos, indica el inicio de la subcadena, incluido en el resultado y el otro indica el final, excluido del resultado. Es importante tener en cuenta que el carácter en la posición final no se incluye en la subcadena.

**Ejemplos:**

```python
# Se crea una variable llamada 'universidad', que almacenará la subcadena del caracter 13 (includo) al 22 (sin incluirlo), de la cadena principal 'mortigo.juan@javeriana.edu.co'.
universidad = email[13:22]
# Se imprime la cadena resultante, almacenada en la variable 'universidad'.
print(universidad)
```

```python
# Se desea obtener el correo estudiantil, excluyendo la parte '@javeriana.edu.co'.
email[0:12]
```

## **Stride**
**Definición propia:** Igual que con el slicing, es la obtención de subcadenas de una cadena original. Sin embargo, para este caso, los caracteres de la subcadena no necesariamente deben estar contiguos en la cadena original. El stride permite extraer carácteres saltando una cantidad específica de posiciones.

Su uso requiere de tres parámetros, separados por el carácter de dos puntos (`:`): `[inicio:final:salto]`. 

El primero es el índice por el cual se comenzará la subcadena (incluyéndolo). El segundo es el índice que va a terminar la subcadena (sin incluirlo). Finalmente, el tercer parámetro, sería la frecuencia con la que se va a tomar un carácter de la cadena original.


**Ejemplos:**

```python
# De la cadena original 'javeriana', se extraen cada dos caracteres. Comenzando desde el índice 0 (incluyéndolo) y terminando en el índice 8 (sin incluirlo).
universidad[0:8:2]
```

```python
# Cuando no hay valor en alguno de los parámetros, se entiende que se recorrerá la cadena entera:
email[::4]
```

## **Concatenación de cadenas:**
**Definición propia:** Es la combinación de cadenas, formando una cadena resultante cuya longitud es la suma de las longitudes de las cadenas que la formaron. 

Se utiliza mediante el carácter de suma (`+`), separando cada cadena por el operador.

**Ejemplos:**

```python
# Creamos 4 variables que contienen diferentes cadenas para luego poder concatenarlas:
primer_nombre = "Juan"
segundo_nombre = "Pablo"
primer_apellido = "Mórtigo"
segundo_apellido = "Salinas"
```

```python
# Concatenamos las cuatro cadenas previamente declaradas, sin espacios:
nombre_completo_sin_espacios = primer_nombre + segundo_nombre + primer_apellido + segundo_apellido
print(nombre_completo_sin_espacios)
```

```python
# Concatenamos las cuatro cadenas previamente declaradas, pero esta vez con espacio de por medio:
nombre_completo_con_espacios = primer_nombre + " " + segundo_nombre + " " + primer_apellido + " " + segundo_apellido
print(nombre_completo_con_espacios)
```

```python
# Concatenamos solamente el primer nombre con el primer apellido, con espacio de por medio:
primer_nombre_y_primer_apellido = primer_nombre + " " + primer_apellido
print(primer_nombre_y_primer_apellido)
```

```python
# Concatenamos una cadena sin variable con una variable anteriormente declarada:
print("Nombre completo:" + " " + nombre_completo_con_espacios)
```

```python
# Mostrar en pantalla una cadena múltiples veces:
nombre_completo_con_espacios * 5
```

## **Secuencias de Escape:**
**Definición propia:** Son maneras de representar formatos dentro de una misma cadena. Se representan siempre con la barra invertida: (`\`) y pueden estar o no acompañadas de otro carácter (como el carácter `n` o `t`).

**Ejemplos:**

```python
# Salto de línea mediante el uso de la secuencia de escape '\n':
print("Nombre completo: \nJuan Pablo Mórtigo Salinas")
```

```python
# Formato de tabulación mediante el uso de la secuencia de escape '\t':
print("Juan Pablo \t Mórtigo Salinas")
```

```python
# Para que la barra invertida forme parte de la cadena, se coloca doble barra invertida:
print("Juan Pablo \\ Mórtigo Salinas")
```

```python
# El ejemplo anterior también funciona colocando el carácter 'r' antes de la cadena y así se muestra la cadena tal cual:
print(r"Juan Pablo \ Mórtigo Salinas") 
```

## **Operaciones con cadenas:**
**Definición propia:** Son métodos que alteran una cadena original. Se realizan mediante funciones predeterminadas.

A todas ellas se accede mediante el carácter `.`.

**Métodos importantes:**
- `upper()`: Convierte a letras mayúsculas la cadena a la que se le aplicó la función.
- `replace()`: Reemplaza un tramo de la cadena por otro asignado.
- `find()`: Encuentra una subcadena dentro de una cadena original, dada dicha subcadena.

**Ejemplos:**

```python
# Convertir el primer nombre 'Juan' en mayúsculas completamente.
primer_nombre_mayuscula = primer_nombre.upper()
print(primer_nombre_mayuscula)
```

```python
# Reemplaza donde se encuentre la palabra "Pablo", por "Felipe".
nombre_completo_modificado = nombre_completo_con_espacios.replace("Pablo", "Felipe")
print(nombre_completo_modificado)
```

```python
# Busca la subcadena "Sali" dentro de la cadena "Juan Felipe Mórtigo Salinas". Devuelve el índice el primer carácter cuando encuentra la subcadena.
nombre_completo_modificado.find("Sali")
```

```python
# Se retorna -1 si, tras recorrer toda la cadena original, no se encuentra la subcadena.
nombre_completo_modificado.find("Sergio")
```

## **Conclusiones:**
1. Las cadenas son secuencias de caracteres que se representan mediante comillas dobles (`" "`) o simples (`' '`).
2. Para mostrar alguna cadena en pantalla, se emplea el método `print()` y para conocer la longitud de una cadena, se utiliza `len()`.
3. La indexación permite acceder a caracteres específicos. En el caso de la indexación positiva, el primer carácter toma el índice 0, el segundo el índice 1 y se continúa hasta finalizar la cadena, mientras que para la indexación negativa, el último carácter recibe el índice de -1, el penúltimo el índice -2 y así hasta llegar al inicio de la cadena.
4. El slicing permite extraer subcadenas de una cadena originaria, siempre y cuando los caracteres a extraer están contiguos.
5. El stride forma subcadenas de una cadena principal, independiente de que los caracteres estén juntos o no.
6. La concatenación con el carácter `+`, permite unir dos o más cadenas de forma sencilla.
7. Métodos como `upper()`, `find()` y `replace()` son herramientas útiles para manipular cadenas.
