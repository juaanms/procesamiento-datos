## **¿Qué es un bucle?**
**Definición propia:** Un bucle es, en otras palabras, un ciclo. Muchas veces, se requiere realizar la misma operación varias veces mientras se cumple una condición determinada. Los ciclos (o bucles) están hechos precisamente para permitir la repetición de un bloque de código. En Python, existen dos tipos de bucles: el bucle `for` y el bucle `while`.

## **Método range():** 
En Python, el método `range()` genera una secuencia ordenada de números que se utiliza comúnmente en bucles `for`. No crea una lista directamente, sino un objeto `range` que produce los números bajo demanda, haciéndolo más eficiente en memoria. El método puede recibir entre uno a tres parámetros:
- `range(5)`: el método genera una secuencia de cinco números enteros, comenzando desde cero: 0, 1, 2, 3, 4.
- `range(3, 9)`: el método genera una secuencua de seis números enteros, comenzando estrictamente desde el número 3 (incluyéndolo) y finaliza en el 9 (sin incluirlo): 3, 4, 5, 6, 7, 8.
- `range(0, 10, 2)`: el método genera una secuencia de cinco números enteros, comenzando desde 0 (incluyéndolo), hasta el 10 (sin incluirlo) y saltándose un número de por medio: 0, 2, 4, 6, 8.

**Ejemplos:**

```python
range(10)
```

```python
range(2, 7)
```

```python
range(2, 20, 3)
```

## **Bucle `for`:** 
Dado que la estructura de control `for` es también un bucle, permite la ejecución de un bloque de código repetidas veces. El bucle `for` tiene demasiados usos, como por ejemplo, imprimir cada elemento de una lista o hacer operaciones con los elementos de una lista.

**Ejemplos:**

```python
# Una forma de usar el bucle `for`:

# Se crea una lista llamada `colores`, en la que cada elemento es una cadena.
colores = ["rojo", "azul", "verde", "amarillo", "morado", "negro", "blanco"]

# Se crea una variable `N` que almacenará la cantidad de elementos dentro de la lista `colores` (la longitud de la lista).
N = len(colores)

# Se utiliza el bucle `for` para recorrer cada elemento de la lista y se imprime uno por uno:
for i in range(N):
    print(colores[i])
```

```python
# Otra forma de uso del bucle `for`:

# Se imprime una secuencia de números desde el 2 (incluyéndolo) hasta el 20 (sin incluirlo), dando saltos cada 3 números:
for i in range(2, 20, 3):
    print(i)
```

```python
# Otra manera para el bucle `for`:

# Sin necesidad de utilizar índices, el `for` toma cada elemento de la lista `colores` y lo asigna a una variable temporal `color`, que se puede usar dentro del bucle:
for color in colores:
    print(color)
```

```python
# Para modificar elementos dentro de una lista:

for i in range(N):
    print("Antes: color", i, "es", colores[i])
    colores[i] = "beige"
    print("Después: color", i, "es", colores[i])
```

```python
# Usamos el método `enumerate()` para acceder al índice y a los elementos de una lista:

animales = ["perro", "gato", "conejo", "caballo", "vaca", "pato"]

for i, animal in enumerate(animales):
    print(i, animal)
```

## **Bucle `while`:** 
Cuando no sabemos cuántas veces se va a repetur un bloque de código, o cuando la repetición depende de una condición que puede cambiar en cualquier momento, se utiliza el bucle `while`.

Este bucle ejecuta un bloque de código mientras una condición sea verdadera. En el momento en que la condición se vuelve falsa, el bucle se detiene automáticamente.

**Ejemplo:**

```python
# Crear lista llamada `calificaciones`
calificaciones = [3.5, 4.2, 2.2, 5.0, 4.7, 1.2, 0.8, 5.0, 5.0, 3.9]

# Inicializar dos variables enteras, denominadas `i` y `calificacion`
i = 0
calificacion = 0

# Mientras la calificación sea diferente de 0.8 y no se haya recorrido toda la lista:
while(calificacion != 0.8):
    # `calificacion` tomará el valor que está almacenado en el índice `i` de la lista `calificaciones`:
    calificacion = calificaciones[i]
    # Continuamos al siguiente índice:
    i = i + 1
    # Imprimimos el valor de la variable `calificacion`
    print(calificacion)

# Imprimimos mensaje en caso de encontrar la calificación de 0.8
if calificacion == 0.8:
    print("\nSe encontró la calificación 0.8")
else:
    print("\nNo se encontró la calificación 0.8 en la lista")
```

## **Examen sobre bucles** 
- Escribe un bucle `for` que imprima los elementos entre -5 y 5 usando la función `range()`.



```python
for i in range(-5, 6):
    print(i)
```

- Imprime los elementos de la siguiente lista: `Genres: ['rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']` y asegúrate de seguir las convenciones del lenguaje Python.

```python
Genres = ["rock", "R&B", "Soundtrack", "R&B", "soul", "pop"]
N = len(Genres)

for i in range(N):
    print(Genres[i])
```

- Escribe un bucle `for` que imprima la siguiente lista: `squares: ['red', 'yellow', 'green', 'purple', 'blue']`

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']
long = len(squares)

for i in range(long):
    print(squares[i])
```

- Escribe un bucle `while` que muestre los valores de la calificación de las canciones de un disco almacenados en la lista `PlayListRatings`. Si la calificación es menor a 6, sal del bucle. La lista `PlayListRatings` está compuesta por: `PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]`

```python
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]

i = 0
rating = 10

while(rating > 6):
    rating = PlayListRatings[i]
    i = i + 1
    print(rating)
```

- Escribe un bucle `while` que copie las cadenas `'orange'` de la lista `squares` a la lista `new_squares`. Detente y sal del ciclo si el valor en la lista no es `orange`:

```python
squares2 = ["orange", "orange", "purple", "blue", "orange"]
new_squares = []

i = 0

while((i < len(squares2)) and (squares2[i] == "orange")):
    new_squares.append(squares2[i])
    i = i + 1

print("Lista original:", squares2)
print("Nueva lista:", new_squares)
```

## **Conclusiones:**
1. Un bucle es una estructura de control que permite la repetición de un determinado bloque de código mientras se cumpla con una o más condiciones.
2. Existen dos tipos de bucles en Python: el bucle `for` y el bucle `while`.
3. La diferencia entre el `for` y el `while` radica en que el `while` es utilizado cuando no se sabe exactamente cuántas repeticiones de un bloque de código ocurrirán, mientras que en el `for`, es necesario indicar la cantidad de repeticiones.
4. La función `range()` genera una secuencia de números enteros ordenados y es altamente utilizada en los ciclos `for`.
