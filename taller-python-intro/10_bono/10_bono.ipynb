1. **lesser_of_two_events**:
"Devuelve el menor si ambos son pares, si no, el mayor"

```python
def lesser_of_two_evens(a,b):
    if a % 2 == 0 and b % 2 == 0:
        # Si ambos números son pares, retorna el menor:
        return min(a,b)
    else:
        # Si al menos uno de los números no es par, retorna el mayor:
        return max(a,b)

print(lesser_of_two_evens(2, 4))
print(lesser_of_two_evens(1, 5))
```

2. **animal_crackers:**
"True si ambas palabras empiezan con la misma letra"

```python
def animal_crackers(text):
    # Separa cada palabra como un elemento de una lista:
    palabras = text.split()
    # Compara la primera letra de cada palabra 
    return palabras[0][0] == palabras[1][0]

print(animal_crackers('Ciudad Calor'))
print(animal_crackers("Lugar Frio"))
```

3. **makes_twenty:**
"True si la suma es 20 o si alguno es 20"

```python
def makes_twenty(n1,n2):
    # Comparar si alguno de los argumentos es 20, o si la suma entre los números ingresados es 20:
    return n1 + n2 == 20 or n1 == 20 or n2 == 20

print(makes_twenty(2, 3))
print(makes_twenty(12, 8))
```

4. **old_macdonald:** "Hacer mayúscula la primera y la cuarta letra"

```python
def old_macdonald(name):
    # Si la longitud del argumento es igual o mayor que 4.
    # Para poder modificar la posición 3 (cuarta letra), la cadena debe tener como mínimo 4 letras.
    if len(name) >= 4:
        #Retorna la primera letra en mayúscula, deja tal cual las letras de la segunda a la tarcera posición (segunda y tercera letra), la letra en la tercera posición (cuarta letra) en mayúscula y todo lo que siga después lo deja con minúsculas:
        return name[0].upper() + name[1:3] + name[3].upper() + name[4:]
    else:
        # Si la cadena es de longitud menor a 4, entonces no se puede modificar y retorna el mismo parámetro.
        return name

print(old_macdonald("macdonald"))
```

5. **master_yoda:**
"Invertir el orden de las palabras"

```python
def master_yoda(text):
    # Convierte el argumento en una lista, en la que cada elemento es una palabra distinta:
    palabras = text.split()
    # Slicing desde el inicio, hasta el final. -1 indica que se recorre la lista de atrás hacia adelante:
    invertida = palabras[::-1]
    # `" "` es un espacio en blanco y `.join(invertida)` une todos los elementos de la lusta usando el espacio en blanco:
    return " ".join(invertida)

print(master_yoda("I am home"))
print(master_yoda("We are ready"))
```

6. **has_33:** "True si hay un 3 seguido de otro 3"

```python
def has_33(nums):
    # Recorrdemos la lista hasta el penúltimo elemento.
    # `len(nums) - 1` se usa para comparar cada elemento con el siguiente en la lista:
    for i in range(len(nums) - 1):
        # Verificar si el elemento actual y el siguiente son 3. Si ambos son 3, entonces se cumple lo que queremos:
        if nums[i] == 3 and nums[i+1] == 3:
            return True
    # Si se recorre toda la lista sin encontrar un par de 3s, entonces no hay 3 consecutivos:
    return False

print(has_33([1, 3, 3]))
print(has_33([1, 3, 1, 3]))
print(has_33([3, 1, 3]))
```

7. **paper_doll:**
"Cada carácter se repite 3 veces"

```python
def paper_doll(text):
    # Crear cadena vacía para construir el resultado
    resultado = ""
    # Recorrer cada letra de la cadena original. La variable `letra` tomará el valor de cada carácter, uno por uno:
    for letra in text:
        # Repetir el carácter 3 veces y agregar al resultado mediante concatenación
        resultado += letra * 3
    # Retorna la cadena del resultado, con todas las letras repetidas 3 veces.
    return resultado

print(paper_doll("Hello"))
print(paper_doll("Mississippi"))
```

8. **blackjack:**
"Suma con reglas de blackjack"

```python
def blackjack(a,b,c):
    # Suma los tres números ingresados como argumento:
    suma = a + b + c
    # Si la suma no pasa de 21, es una mano válida.
    if suma <= 21:
        return suma
    # Si la suma pasa de 21, pero hay un 11 (As), lo convertimos en 1, restando 10 a la suma total.
    elif suma > 21 and (a == 11 or b == 11 or c == 11):
        return suma - 10
    # Si la suma pasa de 21 y no hay un 11 para salvar la mano.
    else:
        return "BUST"

print(blackjack(5,6,7))
print(blackjack(9,9,9))
print(blackjack(9,9,11))
```
