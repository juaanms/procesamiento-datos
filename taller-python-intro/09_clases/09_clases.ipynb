## **¿Qué es una clase?**
**Definición propia:** Una clase es un "molde" para crear objetos. Define un conjunto de características (llamadas atributos) y métodos (denominados comportamientos) que tendrán todos los objetos creados a partir de ella.

## **Conceptos clave:**
- **Clase:** El molde que define cómo serán los objetos.
- **Objeto:** Una instancia de la clase, (lo creado a partir del molde).
- **Atributo:** Una característica del objeto.
- **Método:** Una acción que el objeto puede realizar.

**Ejemplo:**

````python
# Ejemplo de clase y objetos:

# La clase es el molde:
class Estudiante:
    # Constructor para inicializar los atributos:
    def __init__(self, nombre, ID, carrera):
        self.nombre = nombre
        self.ID = ID
        self.carrera = carrera

# Los objetos son las instancias:
estudiante1 = Estudiante("Juan Pablo Mórtigo Salinas", "00020537213", "Ciencia de Datos")
estudiante2 = Estudiante("María Lucía Rodríguez Ovalle", "00020804278", "Arquitectura")
estudiante3 = Estudiante("Sergio Andrés Cárdenas Velasco", "00020702355", "Medicina")

# Imprimir a cada estudiante:
print(estudiante1.nombre)
print(estudiante2.nombre)
print(estudiante3.nombre)
````

## **Creación de una clase:**
Se debe utilizar la palabra reservada `class`, seguida del nombre de la clase (con la primera letra en mayúscula, siempre) y dos puntos `:`.

**Sintaxis básica:**
```python
class NombreClase:
    # Constructor (inicializa el objeto)
    def __init__(self, parametros):
        self.atributo = parametro
    
    # Método (comportamiento del objeto)
    def nombre_metodo(self):
        # Código del método
        pass
```

**Ejemplo:**

````python
# Crear clase `Localidad`:
class Localidad:
    # Inicializar los atributos con el constructor:
    def __init__(self, nombre, area, poblacion):
        self.nombre = nombre
        self.area = area
        self.poblacion = poblacion

    # Método de la clase, denominado `calcular_densidad`:
    def calcular_densidad(self):
        return self.poblacion / self.area

    # Método de la clase, llamado `mostrar_info`
    def mostrar_info(self):
        print("Localidad:", self.nombre)
        print("Área:", self.area, "km²")
        print("Población:", self.poblacion, "habitantes")
        print("Densidad:", self.calcular_densidad(), "hab/km²")
````

## **Crear objetos (instancias)**

**Definición propia:**
Un objeto es una instancia de una clase. Para crear un objeto, se llama a la clase como si fuera una función, pasando los argumentos que requiere el constructor `__init__`.

**Ejemplo:**

````python
# Instancias de la clase `Localidad`:
usaquen = Localidad("Usaquén", 65.54, 503767)
chapinero = Localidad("Chapinero", 35.78, 139701)
suba = Localidad("Suba", 101.07, 1124692)

# Imprimir instancias con algún atributo:
print(usaquen.nombre)
print(chapinero.poblacion)

# Llamar al método `mostrar_info()` de la clase `Localidad`
usaquen.mostrar_info()
print()
chapinero.mostrar_info()
````

## **El constructor `__init__` y `self`**

**Definición propia:**
- **`__init__`:** Es un método especial que se ejecuta automáticamente cuando se crea un objeto. Se usa para inicializar los atributos del objeto.
- **`self`:** Es una referencia al objeto actual. Permite acceder a los atributos y métodos del objeto dentro de la clase.

**Ejemplo:**

````python
# Crear clase llamada `Película`:
class Pelicula:
    # Con `__init__` inicializamos los atributos de la clase:
    def __init__(self, titulo, director, año):
        # Utilizamos `self` para hacer referencia al objeto actual:
        self.titulo = titulo
        self.director = director
        self.año = año

    # Método `mostrar_info()` de la clase `Pelicula`:
    def mostrar_info(self):
        print("Película:", self.titulo)
        print("Director:", self.director)
        print("Año:", self.año)

# Creamos instancias de la clase `Pelicula`:
pelicula1 = Pelicula("Toy Story 5", "Andrew Stanton", 2026)
pelicula2 = Pelicula("El Conjuro", "James Wan", 2013)
pelicula3 = Pelicula("Anyone but you", "Will Gluck", 2023)

# Llamamos al método `mostrar_info` de cada instancia:
pelicula1.mostrar_info()
print()
pelicula2.mostrar_info()
print()
pelicula3.mostrar_info()
````

## **Métodos de una clase**

**Definición propia:**
Los métodos son funciones que pertenecen a la clase y definen el comportamiento de los objetos. Siempre tienen `self` como primer parámetro para acceder al objeto actual.

**Ejemplo:**

````python
# Clase `Pais`:
class Pais:
    # Inicializar attributos de la clase con `__init__`:
    def __init__(self, nombre, capital, poblacion, area, idioma_oficial, moneda):
        # Para referirnos al objeto actual, usamos `self`: 
        self.nombre = nombre
        self.capital = capital
        self.poblacion = poblacion
        self.area = area
        self.idioma_oficial = idioma_oficial
        self.moneda = moneda

    # Método `densidad_poblacion()` de la clase `Pais`:
    def densidad_poblacion(self):
        return self.poblacion / self.area

    # Método `mostrar_info()` de la clase `Pais`:
    def mostrar_info(self):
        print(self.nombre)
        print("Capital:",self.capital)
        print("Población:",self.poblacion,"habitantes")
        print("Área:",self.area,"km²")
        # Utilizamos la función `round()` para redondear. En este caso, dos decimales:
        print("Densidad:",round(self.densidad_poblacion(),2),"km²")
        print("Idioma oficial:",self.idioma_oficial)
        print("Moneda:",self.moneda)

# Crear instancias de la clase (objetos), con sus atributos:
colombia = Pais("Colombia", "Bogotá", 53000000, 1141748, "Español", "Peso colombiano")
brasil = Pais("Brasil", "Brasilia", 215000000, 8515767, "Portugués", "Real")
japon = Pais("Japón", "Tokio", 125000000, 377975, "Japonés", "Yen")
turquia = Pais("Turquia", "Ankara", 85825000, 785350, "Turco", "Lira turca")
india = Pais("India", "Nueva Delhi", 1441720000, 3287263, "Hindi, Inglés", "Rupia india")

# Llamamos al método `mostrar_info` de cada objeto creado:
colombia.mostrar_info()
print()
brasil.mostrar_info()
print()
japon.mostrar_info()
print()
turquia.mostrar_info()
print()
india.mostrar_info()
print()
````

# **Examen sobre clases:**

La clase `matplotlib.patches.Ellipse`, permite dibujar elipses desde `matplotlib`. La clase se importa de la siguiente manera:

`from matplotlib.patches import Ellipse`

Se puede dibujar entonces cualquier elipse, dando los argumentos:

`plt.gca().add_patch(Ellipse(xy, width, height, facecolor = color, edgecolor = edgecolor))`

Se requiere crear una clase Elipse, la cual el constructor sea con valores iniciales cualesquiera. Adicionalmente, la clase debe tener los siguientes métodos:

1.- Cambiar el ancho

2.- Cambiar el alto

3.- Cambiar el color de relleno

4.- Cambiar el color de borde

5.- Dibujar

````python
# Importat las librerías necesarias:
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

class Elipse:
    # Constructor que iniializa los atributos de la clase:
    def __init__(self, centro=(0, 0), ancho=4, alto=2, color_relleno="blue", color_borde="black"):
        self.centro = centro
        self.ancho = ancho
        self.alto = alto
        self.color_relleno = color_relleno
        self.color_borde = color_borde

    # Método para cambiar el ancho:
    def cambiar_ancho(self, nuevo_ancho):
        self.ancho = nuevo_ancho
        print("Ancho cambiado a",self.ancho)

    # Método para cambiar el alto:
    def cambiar_alto(self, nuevo_alto):
        self.alto = nuevo_alto
        print("Alto cambiado a",self.alto)

    # Método para cambiar el color de relleno:
    def cambiar_color_relleno(self,nuevo_color_relleno):
        self.color_relleno = nuevo_color_relleno
        print("Color de relleno cambiado a",self.color_relleno)

    # Método para cambiar el color del borde:
    def cambiar_color_borde(self,nuevo_color_borde):
        self.color_borde = nuevo_color_borde
        print("Color del borde cambiado a",self.color_borde)

    # Método para dibujar la elipse:
    def dibujar(self):
        fig, ax = plt.subplots()
        elipse = Ellipse(self.centro, self.ancho, self.alto, facecolor=self.color_relleno, edgecolor=self.color_borde)
        ax.add_patch(elipse)
        ax.set_xlim(-6,6)
        ax.set_ylim(-4,4)
        ax.set_aspect("equal")
        plt.show()

# Crear y dibujar la elipse (instancia de la clase)
mi_elipse = Elipse(centro=(0, 0), ancho=4, alto=2, color_relleno="green", color_borde="red")
mi_elipse.dibujar()

# Modificar la elipse
mi_elipse.cambiar_ancho(6)
mi_elipse.cambiar_color_relleno("yellow")
mi_elipse.cambiar_color_borde("black")
mi_elipse.dibujar()
````

# **Conclusiones:**
1. Una clase es un molde para crear objetos. Define atributos y métodos.
2. Un objeto es una instancia de una clase, con sus propios atributos y métodos.
3. El constructor `__init__` se ejecuta automáticamente al crear un nuevo objeto y sirve para inicializar sus atributos.
4. `self` es una referencia al objeto actual y se usa dentro de la clase para acceder a sus atributos y métodos.
5. Los métodos son funciones que definen el comportamiento de los objetos.
6. Este paradigma de programación, llamado "Programación Orientada a Objetos" (POO), permite organizar el código de manera más eficiente, legible y reutilizable.
